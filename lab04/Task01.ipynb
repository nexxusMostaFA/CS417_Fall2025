```python
import numpy as np 
```

```python
def step(x):
    return 1 if x >= 0 else 0
```

```python
def neuron(x , w , b):
    result = np.dot(x , w) + b
    return step(result)
```

```python
def compute_gate(dic , gate_type):
     """
    x_dict: { var_name: (value, is_positive) }
    gate_type: "and" or "or"
    """
     values = [ i for i , _ in dic.values()]
     values = np.array(values)

     wights = [1 if i else -1 for _,i  in dic.values()]
     wights = np.array(wights)

     L = sum([i for _,i in dic.values()])

     N = len(values)
     
     b = -L if gate_type == "and" else -(L - N + 1)
     return  neuron(values , wights , b)
```

## (¬y ∧ ¬z) ∨ (¬w ∧ x ∧ ¬y) ∨ (¬x ∧ ¬z):


```python
def test():

    for w in range(2):
        for x in range(2):
            for y in range(2):
                for z in range(2):
                    p1 = {"y:" : (y , False) , "z":(z , False)}
                    p2 = {"w:" : (w , False) , "x":(x , True) , "y":(y , False)}
                    p3 = {"x:" : (x , False) , "z":(z , False)}

                    P4 = {

                        "p1" : (compute_gate(p1 , "and") , True)
                        , "p2" : (compute_gate(p2 , "and") , True)
                        , "p3" : (compute_gate(p3 , "and") , True)
                    }

                    result = compute_gate(P4 , "or")

                    excepted = ( (not y and not z) or (not w and x and not y) or (not x and not z) )

                    print("w   x   y    z")
                    print("---------------")
                    print(f"{w}   {x}   {y}    {z}")
                    print("Result: ", result ,  "     " , "Excepted: ", int(excepted)) 


test()
```

```python
def xor_basic(x):
    n1 = neuron(x , np.array([1 , -1]) , -1)
    n2 = neuron(x , np.array([-1 , 1]) , -1)
    n3 = neuron(np.array([n1 , n2]) , np.array([1 , 1]) , -1)
    return n3
```

```python
def xor_divide_and_conquer(x):
    if len(x) == 1:
        return x[0]
    if len(x) == 2:
        return xor_basic(x)
    mid = len(x) // 2 
    left = x[:mid]
    right = x[mid:]
    left_xor = xor_divide_and_conquer(left)
    right_xor = xor_divide_and_conquer(right)
    return xor_basic(np.array([left_xor , right_xor]))
```

```python
def test_xor_basic():
    print("w   x   y    z")
    print("---------------")
    for w in range(2):
        for x in range(2):
            for y in range(2):
                for z in range(2):
                    val = [x , y , z , w]
                    while len(val) > 1:
                        new_val = [val.pop(0) , val.pop(0)]
                        result = xor_basic(np.array(new_val))
                        val.insert(0 , result)
                    excepted = (x ^ y ^ z ^ w)
                    print(f"{w}   {x}   {y}    {z}" ,{val[0]} ,  "Excepted: ", int(excepted))


                    
test_xor_basic()
```

```python
def test_xor_basic():
    print("w x y z | perceptron expected")
    print("--------------------------")
    for w_val in [0, 1]:
        for x_val in [0, 1]:
            for y_val in [0, 1]:
                for z_val in [0, 1]:
                    ls = [w_val, x_val, y_val, z_val]
                    while len(ls) > 1:
                        vals = [ls.pop(0), ls.pop(0)]
                        out = xor_basic(np.array(vals))
                        ls.insert(0, out)
                    expected = int(x_val ^ y_val ^ z_val ^ w_val)

                    print(f"{w_val} {x_val} {y_val} {z_val} |    {ls[0]}       {expected}")
                    
test_xor_basic()
```

```python
def test_xor_divide_and_conquer():
    print("w x y z | perceptron expected")
    print("--------------------------")
    for w_val in [0, 1]:
      for x_val in [0, 1]:
         for y_val in [0, 1]:
            for z_val in [0, 1]:
               ls = [w_val, x_val, y_val , z_val]
               out = xor_divide_and_conquer(ls)
               expected = int(w_val ^ x_val ^ y_val ^ z_val)
               print(f"{w_val} {x_val} {y_val} {z_val} |    {out}       {expected}")

test_xor_divide_and_conquer()
            
```
