# import necessary libraries 

```python
import numpy as np
from itertools import product 
```

# define step function

```python
def step(x):
    return 1 if x >= 0 else 0
```

# define nueron function 

```python
def nueron(x , w , b):
    return step(x@w + b)
```

# define or gate logic

```python
def or_gate(x):
    result = nueron(x , np.array([1,1]) , -1)
    return result
```

# define and gate logic

```python
def and_gate(x):
    result = nueron(x , np.array([1,1]) , -2)
    return result
```

# define xor gate logic 

```python
def xor_gate_method_1(x):
    n1 = nueron(x , np.array([-1,1]) , -1)
    n2 = nueron(x , np.array([1,-1]) , -1)
    n3 = or_gate(np.array([n1 , n2]))
    return n3 
```

```python
def xor_gate_method_2(x):
    n1 = nueron(x , np.array([1,1]) , -1)
    n2 = nueron(x , np.array([-1,-1]) , 1)
    n3 = nueron(np.array([n1,n2]) , np.array([1,1]) , -2)
    return n3
```

# define multi layer perceptron logic

```python
# (a!xz or a!y) and (xy or !x!z)   ---> [a , x , y , z] ---> [0 ,1 ,2 ,3 ]  
def mlp(x):
    n1 = nueron(x[[0, 1, 3]], np.array([1, -1, 1]), -2)
    n2 = nueron(x[[0, 2]], np.array([1, -1]), -1)
    n3 = nueron(x[[1, 2]], np.array([1, 1]), -2)
    n4 = nueron(x[[1, 3]], np.array([1, 1]), -2)
    n5 = or_gate(np.array([n1, n2]))
    n6 = nueron(np.array([n3, n4]), np.array([1, -1]), 0)
    n7 = and_gate(np.array([n5, n6]))
    return n7
```

# define multi layer perceptron logic for testing 

```python
def mlp_test(x):
    return ((x[0] and not x[1] and x[3]) or (x[0] and not x[2])) and ((x[1] and x[2]) or not (x[1] and x[3]))
```
