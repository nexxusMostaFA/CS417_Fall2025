## Problem2

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
first50_primes = [i for i in range(1000) if is_prime(i)]
```

```python
first50_primes[:50]
```
