```python
def problem1(l, num):
    l.sort() 
    result = []
    left, right = 0, len(l)-1
    while left < right:
        s = l[left] + l[right]
        if s == num:
            result.append((l[left], l[right]))
            left += 1
            right -= 1
        elif s < num:
            left += 1
        else:
            right -= 1
    return result


print(problem1([1, 2, 3, 4, 5, 6], 7))
```

```python
def problem2 (lst):
    freq = {}
    for i in lst:
        result = lst.count(i)
        freq[i] = result
    return freq

print(problem2([1, 2, 2, 3, 3, 3, 4, 4, 4, 4]))  
```
